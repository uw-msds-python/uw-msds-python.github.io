# Python Class Inheritance

In our last section, we learned how object-oriented programming works and how to define our own classes (blueprints) for new objects. We defined a `Pet` object, which we'll continue to use for our discussion of inheritance:

````python
class Pet:
    _next_license_id = 1
    
    def __init__(self, name, category, owner, age):
        self._name = name
        self._category = category
        self._owner = owner
        self._age = age
        self._id = Pet._next_license_id
        Pet._next_license_id += 1

    # Getters
    def get_name(self):
        return self._name

    def get_age(self):
        return self._age

    def get_owner(self):
        return self._owner

    def get_category(self):
        return self._category

    def get_license(self):
        return f'{self._id:05}'

    def full_name(self):
        return f"{self._name} {self._owner.split(' ')[-1]}"

    # Setters
    def rename(self, new_name):
        self._name = new_name

    def set_owner(self, new_owner):
        self._owner = new_owner

    # Dunder methods
    def __repr__(self):
        return f"Pet[{self._name} ({self._category}, age {self._age})]"

    def __lt__(self, other):
        return self._name < other._name 

    def __eq__(self, other):
        return (
            self._name == other._name and
            self._category == other._category and
            self._owner == other._owner and
            self._age == other._age
        )
````

## Real-life hierarchies

Our discussion of classes began with an explanation of real-life objects:

> - A **car** is a type of vehicle that can be used to get around a city (a "class").
>     - The Subaru factory has a blueprint of a car that they use to manufacture more.
> - **My car** is the green Subaru Impreza that I use to get around (an "object", or an instance of a car).
>     - My car can transport **me** (another object) to UW for class (interaction)

Like in real life, however, objects have more complicated relationships than just interacting with each other. There often exist hierarchical, "is-a" relationships between them:
* A Subaru Impreza is a type of car
* A car is a type of powered vehicle
* A powered vehicle is a type of vehicle

These hierarchies are valuable because the more-specific types automatically have behaviors of the more-general types:

* All vehicles have a maximum speed. My Subaru Impreza is a vehicle, so it also has a maximum speed.
* All powered vehicles have a fuel indicator. My Subaru Impreza is a powered vehicle, so it also has a fuel indicator.
* All cars have a steering wheel. My Subaru Impreza is a car, so it also has a steering wheel.

The same type of hierarchy exists in object-oriented programming, and we call it...

## Inheritance

In object-oriented programming, we can model hierarchies with class **inheritance**.

- **Parent classes** (or **superclasses**) define an initial set of data (fields) and behavior (methods).
- **Child classes** (or **subclasses**) are derived from one (or more) parent class(es) and **inherit** their data & behavior
- Child classes can **add to/modify** their parent's data & behavior.
- The inheritance relationship can continue for as many generations of classes as you'd like.

Let's take a look at how this actually works in Python, and see it in action.

While our `Pet` class is somewhat useful, it turns out there are many different varieties of pets that we might want to model distinctly:
- A cat *is a* type of pet
    - A cat has up to 9 lives, whereas regular pets have only one
- A dog *is a* type of pet
    - A dog can fetch, whereas regular pets do not
- A rabbit *is a* type of pet
    - Rabbits reproduce very quickly - maybe they should be addable with the `+` operator

Let's start with the cat. Remember that we can define a new class `Cat` like so:

````python
class Cat:
    pass
````

However, this does not define the inheritance relationship that we described above. To specify that `Cat` should **inherit** from `Pet`, we add the name of the parent class in parentheses:

````python
class Cat:
    pass
````

Although we have not defined any methods, or even a constructor, as part of the `Cat` class, it will automatically inherit all functionality from `Pet`, which means we can create one and use it just like a `Pet`:

````python
cat = Cat('Hobbes', 'Cat', 'Alice Wonderland', 14)
print(cat.full_name())
````

How does Python actually do this? We can gain a bit of insight by asking Python for help:

````python
help(Cat)
````

We can see the inheritance hierarchy in the `Method resolution order`:
* Cat
* Pet
* object

Method resolution order means that when we call a method on an object instance, we look up the method in that order:
* Does the `Cat` class have a `full_name` method? No
* Does the `Pet` class have a `full_name` method? Yes! so we'll use it.

We can also see a list of all the methods that `Cat` inherited from `Pet`.

Lastly, we see in the method resolution order that we're also resolving methods from a class called `object`. This is the true root class of all other classes in Python, and even though we didn't specity a parent type to our `Pet` class, it automatically received `object` as its parent:

````python
# class Pet:
#    ...

# is equivalent to

# class Pet(object):
#    ...
````

## Rethinking the Pet base class

We've defined two subclasses of `Pet`, but is there anything could we do better with the design of our three classes?

?


Let's fix these design issues with `Pet`!

- We'll reduce the initializer redundancy
- We'll add a `speak` to `Pet` to express that all pets should have to define their vocalizations
- We need to figure out a way to prevent creation of the base class.

For the last point, we can take a look at the [ABC (abstract base class) module](https://docs.python.org/3/library/abc.html) and the [@abstractmethod](https://docs.python.org/3/library/abc.html#abc.abstractmethod) decorator.

````python
class Pet:
    _next_license_id = 1
    
    def __init__(self, name, category, owner, age):
        self._name = name
        self._category = category
        self._owner = owner
        self._age = age
        self._id = Pet._next_license_id
        Pet._next_license_id += 1

    # Getters
    def get_name(self):
        return self._name

    def get_age(self):
        return self._age

    def get_owner(self):
        return self._owner

    def get_category(self):
        return self._category

    def get_license(self):
        return f'{self._id:05}'

    def full_name(self):
        return f"{self._name} {self._owner.split(' ')[-1]}"

    # Setters
    def rename(self, new_name):
        self._name = new_name

    def set_owner(self, new_owner):
        self._owner = new_owner

    # Dunder methods
    def __repr__(self):
        return f"Pet[{self._name} ({self._category}, age {self._age})]"

    def __lt__(self, other):
        return self._name < other._name 

    def __eq__(self, other):
        return (
            self._name == other._name and
            self._category == other._category and
            self._owner == other._owner and
            self._age == other._age
        )
````
