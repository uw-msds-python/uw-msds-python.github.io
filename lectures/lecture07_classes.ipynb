### Why object-oriented programming?

For new programmers, structuring code as interactions between objects is often not intuitive, and it takes practice to become good at it. Object-oriented programming is popular because it allows us to logically group *data* with *behavior*.

In procedural programming, you might define a function to operate on a bunch of variables. But there is no code ensuring that they are all being used properly - passed into the function all together, or not mixed and matched with variables for a different computation.

In OOP, we don't have to worry about that because each object's data is stored right next to all the functions (methods) that can be used to interact with it.

You CAN do everything with either OOP or procedural programming, but we often find it very helpful to join the data with behavior rather than the data being separate from the behavior. This makes it very clear which behaviors are associated with which data, and which behaviors you SHOULD use with which data, particularly in a language like Python that does not validate before runtime that you're performing the proper operations with correct types.

## Dunder methods

Above, we added methods to our `Pet` class - an initializer, a method to return the pet's full name, and a method to rename the pet.

But Python actually already provides a number of special methods by default to every object! We saw this with the `__init__` method: if we didn't provide it, we were still able to create a new instance of our class. Python provided a *default* implementation of this special method:

```python
class TestClass:
    pass

test = TestClass()
print(test)
```

These special methods are surrounded by double underscores (like `__init__`) so they are also referred to as **"dunder"** (*double-underscore*) methods.

We can see the motivation for another dunder method when we try to print out an instance of our Dog class:

```python
dunder_dog = Pet('Hapoo', 'Dog', 'Melissa Winstanley', 13)
print(dunder_dog)
```

This works because objects *automatically*, *already*, have a method to translate themselves into a human-readable string. This method, by default, returns the class of the object and the memory location of where the object is stored on the computer.

This is really not a very useful thing to print out when we print our dog object. Instead, we should provide more details specific to a pet. We can do this by defining a special method called `__repr__` to "represent" the object. Typically, it's a good idea to include the class name in the representation.

We call this **overriding**: replacing a default implementation of the method with new behavior.

```python
class Pet:
    def __init__(self, name, category, owner, age):
        self.name = name
        self.category = category
        self.owner = owner
        self.age = age

    def full_name(self):
        return f"{self.name} {self.owner.split(' ')[-1]}"

    def rename(self, new_name):
        self.name = new_name

    # add a function called __repr__ to represent the Pet instance
    

dunder_dog = Pet('Hapoo', 'Dog', 'Melissa Winstanley', 13)
print(dunder_dog)
```

What's happening under the covers? Python is automatically calling your `__repr__` function when you print the object:

```python
print(dunder_dog)
print(dunder_dog.__repr__())
```

There are a number of potentially useful dunder methods that you might want to add to a class:

| Dunder      | Description |
| ----------- | ----------- |
| `__init__` | Initialize |
| `__repr__` | Represent the instance (for debugging) |
| `__str__` | Represent the instance as a string (for printing) |
| `__eq__` | Makes `==` work |
| `__add__` | Makes `+` work |
| `__iter__`, `__next__` | Makes `for ... in ...` work |
| `__getitem__` | Alernative iteration approach |
| `__bool__` | Determine whether an instance is "truthy" |
| `__lt__` / `__gt__` | Makes sorting and `<`/`>` work |
| `__hash__` | Allows you to use the object as a key in a dictionary |

A comprehensive list of possibilities can be found in the [Python documentation](https://docs.python.org/3/reference/datamodel.html#special-method-names).

As another example: let's say we want to be able to sort a list of pets alphabetically by their name. We can do this by implementing the `__lt__` method:

## OOP best practices

While object-oriented programming can be intuitive once you get the hang of it - humans operate in a world of physical objects, after all - there are a few design principles that will help you build high-quality software.

### Getters/setters and encapsulation

In our `Pet` example, we saw that we can access fields of an instance directly:

```python
pet = Pet('Hapoo', 'Dog', 'Melissa Winstanley', 13)
print(pet.age)
```

#### Getters

This is generally considered bad OOP style. The internal data representation of the object should never be accessed directly. Instead, we should provide "getter" methods that allow you to "get" the value of each piece of data:

```python
class Pet:
    def __init__(self, name, category, owner, age):
        self.name = name
        self.category = category
        self.owner = owner
        self.age = age

    # Add "getter" methods
```

We can then use the getter instead of the variable directly when we want to fetch the data:

```python
pet = Pet('Hapoo', 'Dog', 'Melissa Winstanley', 13)
print(pet.get_age())
```

Why is this recommeneded?

Remember that we talked about object design as providing **encapsulation**, or hiding implementation details from users of the class. If the client of the class is using the field directly, then there *is no* encapsulation, which makes changes to the implementation cause client code to break.

Let's say that we realized that "age" is a bad name for this piece of data - what is the unit of age? Is it months, days, years? We want to rename the field to `age_years` to make it clear.

```python
class Pet:
    def __init__(self, name, category, owner, age):
        self.name = name
        self.category = category
        self.owner = owner
        # Rename to age_years
        self.age = age

    def get_age(self):
        return self.age
```

If your client code is using the field directly, then it will break:

```python
pet = Pet('Hapoo', 'Dog', 'Melissa Winstanley', 13)
print(pet.age)
```

By using a getter method instead of a field, we allow the client code to still work, even if the implementation changes:

```python
print(pet.get_age())
```

The renaming of the age variable may seem contrived, but this is very important for more complex classes that may have a lot of sensitive, unintuitive data.

#### Setters

In addition to getters, we also typically provide "setter" methods to set the values of fields (ie `set_fieldname`). Our `rename` method was an example of this!

#### "Private" fields

These getters and setters are all well and good, but they don't actually PREVENT client code from accessing the fields directly:

```python
pet.owner = 'Genghis Khan'
print(pet.owner)
```

Python has no way to prevent this. Some other languages have the ability to mark data as **"private"**, which makes it impossible for client code to access it.

The solution in Python is to use a naming convention to indicate "clients should NOT access this data". The convention for "private" fields is to prefix them with an underscore (`_`). Let's try that now:

```python
# Mark the fields as "private"
class Pet:
    def __init__(self, name, category, owner, age):
        self.name = name
        self.category = category
        self.owner = owner
        self.age = age

    # Getters
    def get_name(self):
        return self.name

    def get_age(self):
        return self.age

    def get_owner(self):
        return self.owner

    def get_category(self):
        return self.category

    # Setters
    def rename(self, new_name):
        self.name = new_name

    def set_owner(self, new_owner):
        self.owner = new_owner
```

Clients *can* still technically access these "private fields:

```python
pet = Pet('Hapoo', 'Dog', 'Melissa Winstanley', 13)
pet._owner = 'Alice Wonderland'
print(pet._owner)
```

However, it is considered very bad programming design to do so, and there are style checkers that we will discuss later that will fail code that accesses private data.

### Designing fields and objects

There is a whole literature on OOP design principles, which we will not examine in any depth in this course. We only need a few basic design tips to get started with.

1. **Single responsibility principle**: each class should be responsible for a single core job. If you find yourself describing a class with an "and", it is likely doing too many things and should be split into multiple classes.
1. **Use good names**: it makes it much easier for client code to use your class appropriately.
1. **One (nontrivial) class per file**: huge files make it harder to make changes and find what you're looking for. Place them in separate files with clear names to make development easier.
    - Exception: trivial classes that are tied to a larger class, like `ContactDetails` might be tied to a `Resume` class.
1. **Keep fields to a minimum**: if a piece of data can be derived from other data, prefer not to store it as additional data. For example, in our `Pet`, we had a `full_name()` method:

```python
class Pet:
    def __init__(self, name, category, owner, age):
        self._name = name
        self._category = category
        self._owner = owner
        self._age = age

    def full_name(self):
        return f"{self._name} {self._owner.split(' ')[-1]}"
```

We would typically NOT want to store this full name as a separate field:

```python
class Pet:
    def __init__(self, name, category, owner, age):
        self._name = name
        self._category = category
        self._owner = owner
        self._age = age
        # BAD - DO NOT DO THIS!
        self._full_name = f"{self._name} {self._owner.split(' ')[-1]}"

    def full_name(self):
        return self._full_name
```

Why not? It makes modifying a name more difficult, since we have to keep both fields in sync:

```python
class Pet:
    def __init__(self, name, category, owner, age):
        self._name = name
        self._category = category
        self._owner = owner
        self._age = age
        self._full_name = f"{self._name} {self._owner.split(' ')[-1]}"

    def full_name(self):
        return self._full_name

    def rename(self, new_name):
        self._name = new_name
        # This function has to reset the full name as well
        self._full_name = f"{self._name} {self._owner.split(' ')[-1]}"
```

When you're designing a class, consider what is the minimum set of data that you can store to fully represent the object.

*Caveat: there are always exceptions. If a computation is very expensive, we may want to do it only once, when it is reset, rather than every time we need it.*

### Don't over-use it

It can be tempting to use object-oriented design for every aspect of your code, but sometimes it's unnecessary. For example, if you are implementing functionality to compute the hypotenuse of a triangle, we could implement that with a class:

```python
import math

class Hypotenuse:
    def compute(self, a, b):
        return math.sqrt(a**2 + b**2)

print(Hypotenuse().compute(4,7))
```

This is somewhat silly.
- There isn't any state associated with the Hypotenuse class
- The computation is a standard mathematical utility
- Clients (users) of the functionality will likely create a bunch of extra Hypotenuse objects every time they have to do the computation

Luckily, unlike other languages, Python doesn't *require* every function to be embedded within a class. So we can just define the function directly:

```python
def euclidean_distance(a, b):
    return math.sqrt(a**2 + b**2)

print(euclidean_distance(4,7))
```

When you're adding code, think about whether it actually needs a class, or fits as part of an existing class, before creating a new class.

## Class variables

The fields that we've given to our classes so far are what we've called *instance variables*: data variables that are associated with a particular *instance* of a class, or a specific object, like a specific pet's age.

However, Python also supports the idea of **class variables**: data that is shared by all instances of the class, and that they can all interact with.

To demonstrate this, consider our pet example again. We originally started this exercise because we had been looking at pet license data from the City of Seattle. So we need the ability to assign license numbers to pets. No two pets should be assigned the same license number, so that we can distinguish between them.

We can keep track of the next pet license id that we want to assign, and assign the ids one by one.

```python
class Pet:    
    def __init__(self, name, category, owner, age):
        self._name = name
        self._category = category
        self._owner = owner
        self._age = age
```

We can show each pet's license number by providing a getter that formats the license nicely. So a license number of 2 should be formatted as `00002`:

```python
class Pet:
    _next_license_id = 1
    
    def __init__(self, name, category, owner, age):
        self._name = name
        self._category = category
        self._owner = owner
        self._age = age
        self._id = Pet._next_license_id
        Pet._next_license_id += 1

    def get_license(self):
        return f'{self._id:05}'
```

## Putting it all together

Whew! Here's our final `Pet` class:

```python
class Pet:
    _next_license_id = 1
    
    def __init__(self, name, category, owner, age):
        self._name = name
        self._category = category
        self._owner = owner
        self._age = age
        self._id = Pet._next_license_id
        Pet._next_license_id += 1

    # Getters
    def get_name(self):
        return self._name

    def get_age(self):
        return self._age

    def get_owner(self):
        return self._owner

    def get_category(self):
        return self._category

    def get_license(self):
        return f'{self._id:05}'

    def full_name(self):
        return f"{self._name} {self._owner.split(' ')[-1]}"

    # Setters
    def rename(self, new_name):
        self._name = new_name

    def set_owner(self, new_owner):
        self._owner = new_owner

    # Dunder methods
    def __repr__(self):
        return f"Pet[{self._name} ({self._category}, age {self._age})]"

    def __lt__(self, other):
        return self._name < other._name 

    def __eq__(self, other):
        return (
            self._name == other._name and
            self._category == other._category and
            self._owner == other._owner and
            self._age == other._age
        )
```
